```python
import math
from collections import Counter
students = [
    ["S1", 2, 60, 0, 0, 0, "Fail"],
    ["S2", 3, 65, 1, 0, 0, "Fail"],
    ["S3", 6, 80, 1, 1, 1, "Pass"],
    ["S4", 7, 85, 1, 1, 1, "Pass"],
    ["S5", 4, 70, 1, 0, 1, "Fail"],
    ["S6", 8, 90, 1, 1, 1, "Pass"]
]
new_student = [5, 75, 1, 1, 0]
numerical_features = ["Study Hours", "Attendance %"]
binary_features = [
    "Assignment Submitted",
    "Lab Participation",
    "Revision Done"
]
print("=== (a) FEATURES ===")
print("Numerical features:")
print(numerical_features)
print("\nBinary features:")
print(binary_features)
def euclidean_distance(a, b):
    total = 0
    for i in range(len(a)):
        total += (a[i] - b[i]) ** 2
    return math.sqrt(total)

def manhattan_distance(a, b):
    total = 0
    for i in range(len(a)):
        total += abs(a[i] - b[i])
    return total

def minkowski_distance(a, b, p=3):
    total = 0
    for i in range(len(a)):
        total += abs(a[i] - b[i]) ** p
    return total ** (1 / p)

def hamming_distance(a, b):
    """
    Hamming distance using ONLY binary features.
    Returns number of mismatched binary features.
    """
    count = 0

    for i in range(len(a)):
        if a[i] != b[i]:
            count += 1

    return count
X = [student[1:6] for student in students]
y = [student[6] for student in students]
euclidean_distances = []

for i in range(len(X)):
    distance = euclidean_distance(new_student, X[i])
    euclidean_distances.append(distance)

print("\n=== (b) EUCLIDEAN DISTANCE ===")

for i in range(len(X)):
    print(students[i][0], ":", round(euclidean_distances[i], 4))
manhattan_distances = []

for i in range(len(X)):
    distance = manhattan_distance(new_student, X[i])
    manhattan_distances.append(distance)

print("\n=== (c) MANHATTAN DISTANCE ===")

for i in range(len(X)):
    print(students[i][0], ":", round(manhattan_distances[i], 4))
minkowski_distances = []

for i in range(len(X)):
    distance = minkowski_distance(new_student, X[i], p=3)
    minkowski_distances.append(distance)

print("\n=== (d) MINKOWSKI DISTANCE (p=3) ===")

for i in range(len(X)):
    print(students[i][0], ":", round(minkowski_distances[i], 4))
new_binary = new_student[2:5]
hamming_distances = []
for i in range(len(X)):
    student_binary = X[i][2:5]

    distance = hamming_distance(new_binary, student_binary)

    hamming_distances.append(distance)

print("\n=== (e) HAMMING DISTANCE (BINARY FEATURES ONLY) ===")

for i in range(len(X)):
    print(students[i][0], ":", hamming_distances[i])
def get_nearest(distances, k=3):

    # Sort indices according to distance
    sorted_indices = sorted(
        range(len(distances)),
        key=lambda i: distances[i]
    )

    nearest_indices = sorted_indices[:k]

    return nearest_indices
def predict_class(distances, k=3):

    nearest_indices = get_nearest(distances, k)

    nearest_labels = [
        y[i] for i in nearest_indices
    ]

    votes = Counter(nearest_labels)

    prediction = votes.most_common(1)[0][0]

    return nearest_indices, nearest_labels, prediction
indices, labels, prediction = predict_class(
    euclidean_distances, 3
)

print("\n=== EUCLIDEAN K=3 ===")

print("3 Nearest Neighbours:")

for i in indices:
    print(
        students[i][0],
        "Distance =",
        round(euclidean_distances[i], 4),
        "Class =",
        y[i]
    )

print("Neighbour classes:", labels)
print("Prediction:", prediction)
indices, labels, prediction = predict_class(
    manhattan_distances, 3
)

print("\n=== MANHATTAN K=3 ===")

print("3 Nearest Neighbours:")

for i in indices:
    print(
        students[i][0],
        "Distance =",
        round(manhattan_distances[i], 4),
        "Class =",
        y[i]
    )

print("Neighbour classes:", labels)
print("Prediction:", prediction)
indices, labels, prediction = predict_class(
    minkowski_distances, 3
)

print("\n=== MINKOWSKI p=3, K=3 ===")

print("3 Nearest Neighbours:")

for i in indices:
    print(
        students[i][0],
        "Distance =",
        round(minkowski_distances[i], 4),
        "Class =",
        y[i]
    )

print("Neighbour classes:", labels)
print("Prediction:", prediction)
indices, labels, prediction = predict_class(
    hamming_distances, 3
)

print("\n=== HAMMING K=3 ===")

print("3 Nearest Neighbours:")

for i in indices:
    print(
        students[i][0],
        "Distance =",
        hamming_distances[i],
        "Class =",
        y[i]
    )

print("Neighbour classes:", labels)
print("Prediction:", prediction)
print("\n=== (h) FINAL COMPARISON ===")

print("Euclidean  : Pass")
print("Manhattan  : Pass")
print("Minkowski  : Pass")
print("Hamming    : Pass")

print("\nMost suitable distance measure:")
print("Hamming distance is useful when considering ONLY binary features.")
print("For the complete dataset, Euclidean/Manhattan/Minkowski can be used,")
print("but the numerical features should ideally be normalized because")
print("Attendance (%) has a much larger numerical scale than the binary features.")
```
